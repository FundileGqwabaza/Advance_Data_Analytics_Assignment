```python
##Fixing date format for viewership
df_viewership['RecordDate2'] = pd.to_datetime(df_viewership['RecordDate2'], format='%m/%d/%y %H:%M')
##Merging profile & viewership data using left Join
df_user_profile_temp = pd.read_csv("/Workspace/Users/gqwabaza7@gmail.com/Advance_Data_Analytics_Assignment/user_profile_df.csv")
```

```python
##Converting duration to minutes
df_viewership['Duration'] = pd.to_timedelta(df_viewership['Duration 2'])    
df_viewership['Duration'] = df_viewership['Duration']/60
```

```python
##extracting the hour from the date
df_viewership['end_time'] = df_viewership['RecordDate2'] + df_viewership['Duration']
```

```python
##Converting to sast timezone
if df_viewership['RecordDate2'].dt.tz is None:
    df_viewership['RecordDate2'] = df_viewership['RecordDate2'].dt.tz_localize('Africa/Johannesburg')
else:
    df_viewership['RecordDate2'] = df_viewership['RecordDate2'].dt.tz_convert('Africa/Johannesburg')
```

```python
df_viewership['hour'] = df_viewership['RecordDate2'].dt.hour
df_viewership['day'] = df_viewership['RecordDate2'].dt.day_name()
df_viewership['date'] = df_viewership['RecordDate2'].dt.date
df_viewership['month'] = df_viewership['RecordDate2'].dt.month
df_viewership['year'] = df_viewership['RecordDate2']
df_viewership['is_weekend'] = df_viewership['day'].isin(['Saturday','Sunday'])
```

```python
df_viewership['duration 2'] = pd.to_timedelta(df_viewership['Duration'].dt.components.hours, unit='h')+pd.to_timedelta(df_viewership['Duration'].dt.components.minutes, unit='m')+pd.to_timedelta(df_viewership['Duration'].dt.components.seconds, unit='s')
df_viewership.head(10)
```
